## How this repo is set up for Render
- The app to deploy is `tutorial_app/app.py`; it must define `server = app.server`.
- `tutorial_app/requirements.txt` lists **only** what the app needs to run online
  (no Jupyter), which keeps the Render build small. The lesson uses `pip freeze`, but
  that would include all of JupyterLab.
- In Render, set **Root Directory** to `tutorial_app`, **Start Command** to
  `gunicorn app:server`, and the environment variable `PYTHON_VERSION` to the version below.
