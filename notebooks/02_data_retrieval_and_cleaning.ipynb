## Preparing Data for Visualization

```python
def to_df(queryurl):
  response = requests.get(queryurl)
  content_text = StringIO(response.content.decode('utf-8'))
  df = pd.read_csv(content_text)
  return df

df_ukr = to_df(query_url_ukr)
df_rus = to_df(query_url_rus)

df_rus.head()
```
